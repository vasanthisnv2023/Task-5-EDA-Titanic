```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = sns.load_dataset('titanic')
df.head()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['age'], bins=20, kde=True)

plt.title("Age Distribution")
plt.show()
```

Most passengers were between 20 and 40 years old.

```python
plt.figure(figsize=(6,4))

sns.countplot(x='sex', data=df)

plt.title("Gender Distribution")
plt.show()
```

Male passengers were more numerous than females.

```python
plt.figure(figsize=(6,4))

sns.countplot(x='class', data=df)

plt.title("Passenger Class Distribution")
plt.show()
```

Third Class had the highest number of passengers.

```python
plt.figure(figsize=(8,4))

sns.boxplot(y=df['age'])

plt.title("Age Boxplot")
plt.show()
```

A few age outliers are present.

```python
plt.figure(figsize=(8,4))

sns.boxplot(y=df['fare'])

plt.title("Fare Boxplot")
plt.show()
```

Several high-fare outliers are visible.

```python
plt.figure(figsize=(7,5))

sns.countplot(
    x='sex',
    hue='survived',
    data=df
)

plt.title("Survival by Gender")
plt.show()
```

Female passengers had a much higher survival rate.

```python
plt.figure(figsize=(7,5))

sns.countplot(
    x='class',
    hue='survived',
    data=df
)

plt.title("Survival by Passenger Class")
plt.show()
```

First-class passengers survived more frequently than Third-class passengers.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='age',
    y='fare',
    data=df
)

plt.title("Age vs Fare")
plt.show()
```

There is no strong relationship between age and fare.

```python
numeric_df = df.select_dtypes(include=['number'])

plt.figure(figsize=(10,8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.show()
```

Fare shows a positive correlation with survival.
Passenger class is negatively correlated with survival.

```python
sns.pairplot(
    df[['age','fare','pclass','survived']]
)

plt.show()
```

Pairplot helps visualize relationships among multiple variables simultaneously.
